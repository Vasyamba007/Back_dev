# Requests

В requests используются функции с теми же названиями, что и HTTP-запросы: get(), post(), put() и другие.
Рассмотрим пример GET-запроса к Static API Яндекс Карт:

```python
import requests

response = requests.get("https://static-maps.yandex.ru/1.x/?"
               "ll=37.677751,55.757718&"
               "spn=0.016457,0.00619&"
               "l=map")
print(response)
```

Код ответа 200 означает, что запрос был успешно обработан сервером. Теперь можно извлечь полученные данные. Они находятся в атрибуте .content и представлены в виде последовательности байтов — это изображение карты в формате PNG.

У функции get() есть удобный именованный аргумент params, в который можно передать словарь параметров, — это делает код чище и понятнее.

При работе с API важно предусматривать обработку нештатных ситуаций, например отсутствие подключения к интернету. Если Сеть недоступна, requests выбросит исключение ConnectionError.

```python
params = {
    "ll": "37.677751,55.757718",
    "spn": "0.016457,0.00619",
    "l": "map"
}

try:
    response = requests.get("https://static-maps.yandex.ru/1.x/", params=params)
except requests.ConnectionError as e:
    print('Отсутствует подключение к сети')
else:
    with open("images/map.png", "wb") as file:
        file.write(response.content)
```
